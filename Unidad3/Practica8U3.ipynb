```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Valores de la variable
x = [0, 1, 2, 3, 4, 5]
p = [0.05,0.10, 0.25, 0.30, 0.20, 0.10]

#Verificar suma = 1
print ("Suma de probabilidades: ", sum(p))

#Valor esperado E(X)
esperanza = 0
for i in range (len(x)):
  esperanza += x[i]* p[i]

#Calcular E(X^2)
esperanza_cuadrado = 0
for i in range (len(x)):
  esperanza_cuadrado += (x[i] ** 2) * p[i]

#Varianza y desviación estándar
varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)

print (f"Valor esperado = {esperanza:.2f}")
print (f"Varianza = {varianza:.2f}")
print (f"Desviación estandar = {desviacion:.2f}")
```

```python
#Funcion acumulada
F = np.cumsum(p)

for xi, fi in zip (x,F):
  print (f"F({xi}) = {fi:.2f}")

#Graficar
plt.step(x, F, where ='post')
plt.title ("Funcion de Distribución Acumulada (FDA)")
plt.xlabel("Unidades vendidas")
plt.ylabel ("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0,1.05)
plt.show()
```
