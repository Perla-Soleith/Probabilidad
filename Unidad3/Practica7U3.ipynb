```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Valores de la variable
x = [0, 10, 20, 50]
p = [0.50,0.30, 0.15, 0.05]

#Valor esperado E(X)
esperanza = 0
for i in range (len(x)):
  esperanza += x[i]* p[i]

#Calcular E(X^2)
esperanza_cuadrado = 0
for i in range (len(x)):
  esperanza_cuadrado += (x[i] ** 2) * p[i]

#Varianza y desviación estándar
varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)

print (f"Valor esperado = {esperanza:.2f}")
print (f"Varianza = {varianza:.2f}")
print (f"Desviación estandar = {desviacion:.2f}")
```

```python
#Funcion acumulada
F = np.cumsum(p)

for xi, fi in zip (x,F):
  print (f"F({xi}) = {fi:.2f}")

#Graficar
plt.step(x, F, where ='post')
plt.title ("Funcion de Distribución Acumulada (FDA)")
plt.xlabel("Costo")
plt.ylabel ("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0,1.05)
plt.show()
```

**• ¿Qué significa el valor esperado en este contexto?**

Es el promedio de ganancia del juego si se repite muchas veces, se ganan 8.5 del costo por jugada.
