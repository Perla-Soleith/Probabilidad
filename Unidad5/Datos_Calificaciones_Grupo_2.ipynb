```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression # modelo de regresión Lineal
import numpy as np
import seaborn as sns
from scipy.stats import pearsonr
```

```python
df2=pd.read_csv("grupo2.csv")
df2.head()
```

```python
df2.info()
```

```python
df2.describe()
```

```python
#Grafica
plt.hist(df2["Calificacion"], bins=10, color = 'purple', alpha = 0.5)
plt.title("Histograma de Calificaciones")
plt.xlabel("Calificaciones")
plt.ylabel("Frecuencia")
plt.show()
```

#***Regresión Lineal***

```python
X = df2[["HorasEstudio"]]
y = df2["Calificacion"]

modelo = LinearRegression()
modelo.fit(X, y)

#La ecuacion de la línea de regresión

pendiente = modelo.coef_[0]
intercepto = modelo.intercept_

print("Pendiente (m):", pendiente)
print("Intercepto (b):", intercepto)

print(f"Ecuación de la recta:")
print(f"y = {pendiente:.2f}x + {intercepto:.2f}")
```

```python
#Pearson

pearson = df2["HorasEstudio"].corr(df2["Calificacion"])

print("Coeficiente de correlación de Pearson:", pearson)
```

```python
#La grafica

plt.figure(figsize=(8,5))
sns.scatterplot(
    x=df2["HorasEstudio"], y=df2["Calificacion"], color="purple")
plt.plot(df2["HorasEstudio"], modelo.predict(X), color="black")
plt.title("Regresión Lineal: Horas de Estudio vs Calificación")
plt.xlabel("Horas de Estudio")
plt.ylabel("Calificación")
plt.grid(True)
plt.show()
```

#***Coeficiente de Correlación***

```python
horas_estudio = df2["HorasEstudio"]
calificaciones = df2["Calificacion"]

# Calcular coeficiente de correlación de Pearson
coeficiente_correlacion, p_valor = pearsonr(horas_estudio, calificaciones)

# Mostrar resultados
print("Coeficiente de correlación de Pearson:", coeficiente_correlacion)
print("Valor p:", p_valor)
```

#***Resultados:***

**1. ¿Explicar cuál es la interpretación del coeficiente de regresión en términos de la relación entre el tiempo de estudio y las calificaciones finales?**

La pendiente positiva (4.22) indica que por cada hora adicional de estudio, la calificación aumenta aproximadamente 4.22 puntos.


**2. ¿Cómo se interpreta el coeficiente de correlación en términos de la fuerza y dirección de la relación entre las variables?**

El coeficiente de Pearson (0.862) indica una relación positiva y fuerte entre las horas de estudio y las calificaciones.

**3. ¿El modelo de regresión lineal es estadísticamente significativo?**

Sí, porque el valor p es muy pequeño y menor que 0.05.

**4. ¿Qué tan confiable es el modelo de regresión para predecir las calificaciones finales de los estudiantes basado en el tiempo de estudio?**

El modelo es bastante confiable, ya que existe una relación fuerte entre las horas de estudio y las calificaciones.
