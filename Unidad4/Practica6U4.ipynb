```python
from scipy.stats import poisson
```

```python
#Parámetros de la distribución de Poisson
media_accidentes = 2

#Parte 1. Probabilidad de que no haya accidentes en una semana
probabilidad_0_accidentes = poisson.pmf (0, media_accidentes)

print ("1. porbabilidad de queno haya accidentes en 1 semana: ", probabilidad_0_accidentes)

#Parte 2: Probabilidad de que a la suma haya tres accidentes en 2 semanas
#Caonvertimos la media a una media para 2 semanas
media_2_semanas = media_accidentes * 2

#Calculamos la probabilidad acumulada hasta 3 accidentes en 2 semanas
probabilidad_3_o_menos_accidentes = poisson.cdf (3, media_2_semanas)

print ("2. Probabilidad de que a la suma haya tres accidentes en 2 semanas: ", probabilidad_3_o_menos_accidentes)
```
