```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
plt.figure(figsize=(6,4))

#Parámetros de la distribución normal
media = 70 #Media
desv_std = 4 #Desviación estándar
total_estudiantes = 500

#Variable para cálculo de la probabilidad de que un estudiante pese menos de 75 kg
peso_limite = 75

#Primero necesitaos calcular la probailidad de que un estudiante pese menos
#de 75 kg utilizando la función de la distribución acumulativa (CDF) de la distribución normal
prob_menos_75kg = norm.cdf(peso_limite, media, desv_std)

#Cálculo del número de estudiantes que pesan menos de 75 kg
estudiantes_menos_75kg = total_estudiantes * prob_menos_75kg

print ("Probabilidad de que un estudiante pese menos de 75 kg: ", prob_menos_75kg)
print("Numero de estudiantes que pesan menos de 75 kg:", estudiantes_menos_75kg)

#Graficar la distribución normal
x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf (x, media, desv_std)

plt.plot(x, y, 'b', linewidth=2)
plt.fill_between (x, y, where= (x <= peso_limite), color = 'skyblue', alpha = 0.5)
plt.title('Probabilidad de que un estudiante pese menos de 75 kg. \n $\sigma=4$, x=500')
plt.xlabel ('Peso (kg)')
plt.ylabel ('Densidad de probabilidad')
plt.grid(True)
plt.show()
```
