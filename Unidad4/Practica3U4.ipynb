```python
from scipy.stats import binom
```

```python
#Parámetros de la distribucion binomial
probabilidad_exito = 0.83
num_pacientes = 16
X = 16 #Numero de éxitos

#Calcular la probabilidad de que ningún paciente sufra la enfermedad
probabilidad_ninguno_enfermo = binom.pmf (X, num_pacientes, probabilidad_exito)

print ("La probabilida de que ningún paciente sufra la enfermedad es:", probabilidad_ninguno_enfermo)
```

```python
#Parámetros de la distribucion binomial
probabilidad_exito = 0.83
num_pacientes = 16
X = 0

#Calcular la probabilidad de que todos os pacientes sufran la enfermedad
probabilidad_ninguno_enfermo = binom.pmf (X, num_pacientes, probabilidad_exito)

print ("La probabilida de que ningún paciente sufra la enfermedad es:", probabilidad_ninguno_enfermo)
```
