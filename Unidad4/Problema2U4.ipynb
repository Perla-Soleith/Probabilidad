```python
from scipy.stats import poisson
```

El número de X personas ingresadas a una unidad de cuidados intensivos en un hospital particual, en un día , tiene una distribución de probabilidad de Poisson con media igual a 5 personas por día.

a) Cuál es la probabilidad  de que el número de personas ingresadas a una unidad de cuidados intensivos en un hospital particular, sea dos?


```python
media_personas_dia = 5

x = 2

prob_pers_ingres = poisson.pmf (x, media_personas_dia)

print ("La probabilidad de el número de personas ingresadas de cuidados intensivos en un hospital sea dos: ",
       prob_pers_ingres)
```

#b)Es probable que x exceda de 10? Explicar

Si, es probable pero muuuy ligeramente

```python
media_personas_dia = 5
x = 10
prob_pers_ingres = poisson.pmf (x, media_personas_dia)
print ("La probabilidad de el número de personas ingresadas de cuidados intensivos en un hospital sea 10: ",
       prob_pers_ingres)
```
