```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
#Parámetros de la distribucución binomial
n = 10
p = 0.1

#Calcular las probabilidades para cada valor de x de 0 a 10
valores_x = np.arange (0, n+1)
probabilidades_x = [binom.pmf (x, n, p) for x in valores_x]

#Imprimir la tabla de vañlores de x y sus probabilidades
print ("Valores de x\tProbabilidades")
for x, probabilidad in zip(valores_x, probabilidades_x):
  print (f"{x}\t\t{probabilidad}")
```

```python
#Graficar el histograma de la distribución normal
plt.figure (figsize=(8,6))
plt.bar (valores_x, probabilidades_x, color = 'pink')
plt.title ('Distribución Binoimial')
plt.xlabel ('Valor de X')
plt.ylabel ('Probabilidad')
plt.xticks (valores_x)
plt.grid(True)
plt.show
```
