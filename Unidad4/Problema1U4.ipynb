Una revisión de los pesos de 500 estudiantes del ITD, evidenció que la media es de 70 kilogramos y la desviación estándar es de 4 kilogramos. Suponiendo que los valores se distribuyen normalmente. Hallar lo siguiente:

La probabilidad de estudiantes que pesan más de 68 y menos de 72 kg.
Cuantos estudiantes pesan más de 68 y menos de 72 kg.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
plt.figure(figsize=(6,4))

# Parámetros de la distribución normal
media = 70  # Media
desv_std = 4  # Desviación estándar
total_estudiantes = 500

# Límites del intervalo
limite_inferior = 68
limite_superior = 72

# Probabilidades acumuladas
prob_menos_72 = norm.cdf(limite_superior, media, desv_std)
prob_menos_68 = norm.cdf(limite_inferior, media, desv_std)

# Probabilidad entre 68 y 72
prob_entre = prob_menos_72 - prob_menos_68

# Número de estudiantes
estudiantes_entre = total_estudiantes * prob_entre

print("Probabilidad de que un estudiante pese entre 68 y 72 kg:", prob_entre)
print("Número de estudiantes que pesan entre 68 y 72 kg:", estudiantes_entre)

# Graficar la distribución normal
x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf(x, media, desv_std)

plt.plot(x, y, 'black', linewidth=2)
plt.fill_between(x, y, where=(x >= limite_inferior) & (x <= limite_superior),
                 color='purple', alpha=0.5)

plt.title('Probabilidad de que un estudiante pese entre 68 y 72 kg\n$\mu=70$, $\sigma=4$, n=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)

plt.show()
```
